```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Step 1: Data
x = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)   # Study hours (input)
y = np.array([40, 50, 60, 70, 80])              # Marks (output)

# Step 2: Create model
model = LinearRegression()

# Step 3: Train model
model.fit(x, y)

# Step 4: Predict marks for 6 hours of study
predicted = model.predict([[6]])
print("Predicted marks for 6 hours of study:", predicted[0])

# Step 5: Plot the line
plt.scatter(x, y, color='blue', label='Actual data')
plt.plot(x, model.predict(x), color='red', label='Best fit line')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Step 1: Create the DataFrame
data = {
    'Size_sqft': [1000, 1500, 2000, 2500, 3000, 3500],
    'Price': [150000, 200000, 250000, 300000, 350000, 400000]
}

df = pd.DataFrame(data)
print("🏠 House Price Data:")
print(df)

# Step 2: Split into input (X) and output (y)
X = df[['Size_sqft']]   # input feature (2D array)
y = df['Price']         # target variable

# Step 3: Create and train the model
model = LinearRegression()
model.fit(X, y)

# Step 4: Predict the price of a new house (e.g., 4000 sqft)
predicted_price = model.predict([[4000]])
print(f"\nPredicted price for 4000 sqft house: ${predicted_price[0]:,.0f}")

# Step 5: Plot the data and regression line
plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X, model.predict(X), color='red', label='Best Fit Line')
plt.xlabel('House Size (sqft)')
plt.ylabel('Price ($)')
plt.title('House Price Prediction using Linear Regression')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Data
x = np.array([1,2,3,4,5]).reshape(-1,1)
y = np.array([40,50,60,65,80])

# Fit sklearn linear regression
model = LinearRegression()
model.fit(x, y)

print("Slope (m):", model.coef_[0])       # should be 9.5
print("Intercept (c):", model.intercept_) # should be 30.5

# Predictions
y_pred = model.predict(x)
for xi, yi, yp in zip(x.flatten(), y, y_pred):
    print(f"x={xi}, actual={yi}, predicted={yp:.1f}, residual={yi-yp:.1f}")

# MSE
mse = np.mean((y - y_pred)**2)
rmse = np.sqrt(mse)
print("MSE:", mse, "RMSE:", rmse)

# Plot
plt.scatter(x, y, label='Actual')
plt.plot(x, y_pred, label='Regression line')
plt.xlabel('Study hours')
plt.ylabel('Marks')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Data
x = np.array([1000,1200,1400,1600,1800,2000]).reshape(-1,1)
y = np.array([18.0,16.5,15.0,13.5,12.0,10.5])

# Fit model
model = LinearRegression()
model.fit(x, y)

print("Slope (m):", model.coef_[0])       # -0.0075
print("Intercept (c):", model.intercept_) # 25.5

# Predictions and error
y_pred = model.predict(x)
for xi, yi, yp in zip(x.flatten(), y, y_pred):
    print(f"x={xi}, actual={yi}, predicted={yp:.2f}, residual={yi-yp:.2f}")

mse = np.mean((y - y_pred)**2)
rmse = np.sqrt(mse)
print("MSE:", mse, "RMSE:", rmse)

# Plot
plt.scatter(x, y, label='Actual data')
plt.plot(x, y_pred, label='Regression line')
plt.xlabel('Weight (kg)')
plt.ylabel('Mileage (km/l)')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Data
x = np.array([1,2,3,4,5,6]).reshape(-1,1)
y = np.array([35000,38000,42000,46000,52000,60000])

# Fit
model = LinearRegression()
model.fit(x, y)

print("Slope (m):", model.coef_[0])       # ~4885.7143
print("Intercept (c):", model.intercept_) # 28400

# Predictions
y_pred = model.predict(x)
for xi, yi, yp in zip(x.flatten(), y, y_pred):
    print(f"x={xi}, actual={yi}, predicted={yp:.2f}, residual={yi-yp:.2f}")

# Errors
mse = np.mean((y - y_pred)**2)
rmse = np.sqrt(mse)
print("MSE:", mse, "RMSE:", rmse)

# Plot
plt.scatter(x, y, label='Actual')
plt.plot(x, y_pred, label='Regression line')
plt.xlabel('Years of Experience')
plt.ylabel('Salary ($)')
plt.legend()
plt.show()
```
